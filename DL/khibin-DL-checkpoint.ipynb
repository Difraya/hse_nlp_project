Теперь создадим датасет для PyTorch'а.

```python
class DatasetBooks(Dataset):
    def __init__(self, X, y, vectorizer=None, is_train=True, from_sparse_to_array=True):
        super().__init__()
        self.X = X
        self.y = y
        self.vectorizer = vectorizer
        self.is_train = is_train
        self.from_sparse_to_array = from_sparse_to_array

        if vectorizer:
            if self.is_train:
                self.X = self.vectorizer.fit_transform(self.X)
            else:
                self.X = self.vectorizer.transform(self.X)

            if self.from_sparse_to_array:
                to_dense = FunctionTransformer(lambda x: x.toarray())
                self.X = to_dense.fit_transform(self.X)


        self.X, self.y = torch.Tensor(self.X), torch.Tensor(self.y.to_numpy())
        self.shape = X.size

    def __getitem__(self, indx):        
        return self.X[indx], self.y[indx]
    
    def __len__(self):
        return self.X.shape[0]
```

```python
vectorizer = TfidfVectorizer(max_features=5000, ngram_range=(1, 2))

train_ds_tfidf = DatasetBooks(X_train, y_train, vectorizer, is_train=True, from_sparse_to_array=True)
valid_ds_tfidf = DatasetBooks(X_valid, y_valid, vectorizer, is_train=False, from_sparse_to_array=True)
test_ds_tfidf = DatasetBooks(X_test, y_test, vectorizer, is_train=False, from_sparse_to_array=True)

train_loader_tfidf = DataLoader(train_ds_tfidf, batch_size=16, shuffle=True)
valid_loader_tfidf = DataLoader(valid_ds_tfidf, batch_size=16, shuffle=True)
test_loader_tfidf = DataLoader(test_ds_tfidf, batch_size=16, shuffle=True)
```

### Функция для обучения

```python
def train(
    model: torch.nn.Module,
    train_loader: torch.utils.data.DataLoader,
    val_loader: torch.utils.data.DataLoader,
    optimizer: torch.optim.Optimizer,
    loss_fn: Callable,
    metric_fn: Callable,
    epochs: int = 10,
    scheduler: Optional[torch.optim.lr_scheduler._LRScheduler] = None,
    init: Optional[Callable] = None,
    verbose: bool = True,
    verbose_step: int = 1,
    device: str = 'cuda' if torch.cuda.is_available() else 'cpu',
    early_stopping: Optional[int] = None,
    grad_clip: Optional[float] = None,
    checkpoint_path: Optional[str] = None,
    log_fn: Optional[Callable] = None
) -> Dict[str, list]:
    """
    Train a PyTorch model with various options for training control and monitoring.
    """
    if init is not None:
        for module in model.modules():
            if isinstance(module, (nn.Linear, nn.Conv2d)):
                init(module.weight)
                if module.bias is not None:
                    module.bias.data.fill_(0.0)
    
    model.to(device)
    
    history = {
        'train_loss': [],
        'val_loss': [],
        'train_metric': [],
        'val_metric': [],
        'lr': []
    }
    
    best_val_metric = -np.inf
    early_stop_counter = 0
    
    for epoch in range(1, epochs + 1):
        model.train()
        epoch_train_loss = 0.0
        epoch_train_metric = 0.0
        batch_count = 0
        
        with tqdm(train_loader, unit="batch", desc=f"Epoch {epoch}/{epochs}", disable=not verbose) as pbar:
            for batch_idx, (data, target) in enumerate(pbar):
                data, target = data.to(device), target.to(device)

                target = target.long()
                
                optimizer.zero_grad()
                output = model(data)
                
                loss = loss_fn(output, target)
                
                loss.backward()
                
                if grad_clip is not None:
                    torch.nn.utils.clip_grad_norm_(model.parameters(), grad_clip)
                
                optimizer.step()
                
                metric = metric_fn(output, target).item()
                epoch_train_loss += loss.item()
                epoch_train_metric += metric
                batch_count += 1
                
                pbar.set_postfix({
                    'loss': loss.item(),
                    'metric': metric,
                    'lr': optimizer.param_groups[0]['lr']
                })
                
                if log_fn is not None:
                    log_fn({
                        'epoch': epoch,
                        'batch': batch_idx,
                        'batch_loss': loss.item(),
                        'batch_metric': metric,
                        'lr': optimizer.param_groups[0]['lr']
                    })
        
        epoch_train_loss /= batch_count
        epoch_train_metric /= batch_count
        history['train_loss'].append(epoch_train_loss)
        history['train_metric'].append(epoch_train_metric)
        history['lr'].append(optimizer.param_groups[0]['lr'])
        
        model.eval()
        val_loss = 0.0
        val_metric = 0.0
        with torch.no_grad():
            for data, target in val_loader:
                data, target = data.to(device), target.to(device)

                target = target.long()
                
                output = model(data)
                val_loss += loss_fn(output, target).item()
                val_metric += metric_fn(output, target).item()
        
        val_loss /= len(val_loader)
        val_metric /= len(val_loader)
        history['val_loss'].append(val_loss)
        history['val_metric'].append(val_metric)
        
        if scheduler is not None:
            if isinstance(scheduler, torch.optim.lr_scheduler.ReduceLROnPlateau):
                scheduler.step(val_loss)
            else:
                scheduler.step()
        
        if verbose:
            print(f"Epoch {epoch}/{epochs} - "
                  f"Train Loss: {epoch_train_loss:.4f}, Train Metric: {epoch_train_metric:.4f} | "
                  f"Val Loss: {val_loss:.4f}, Val Metric: {val_metric:.4f} | "
                  f"LR: {optimizer.param_groups[0]['lr']:.2e}")
        
        if log_fn is not None:
            log_fn({
                'epoch': epoch,
                'train_loss': epoch_train_loss,
                'train_metric': epoch_train_metric,
                'val_loss': val_loss,
                'val_metric': val_metric,
                'lr': optimizer.param_groups[0]['lr']
            })
        
        if checkpoint_path is not None:
            if val_metric > best_val_metric:
                best_val_metric = val_metric
                torch.save(model.state_dict(), checkpoint_path)
                if verbose:
                    print(f"Model improved (val_metric = {val_metric:.4f}). Saved to {checkpoint_path}")
        
        if early_stopping is not None:
            if val_metric > best_val_metric:
                best_val_metric = val_metric
                early_stop_counter = 0
            else:
                early_stop_counter += 1
                if early_stop_counter >= early_stopping:
                    if verbose:
                        print(f"Early stopping at epoch {epoch} (no improvement for {early_stopping} epochs)")
                    break
    
    return history
```

```python
# Функция для отрисовки графиков
def plot_loss_metrics(train_loss, val_loss, train_metric, val_metric, figsize=(10, 5)):
    sns.set_theme(rc={'figure.figsize': figsize})
    
    sns.lineplot(data=train_loss);
    sns.lineplot(data=val_loss);
    plt.title("Функция потерь");
    plt.xlabel('Итерации');
    plt.ylabel('Функция потерь');
    plt.show();

    sns.lineplot(data=train_metric);
    sns.lineplot(data=val_metric);
    plt.title("Метрика");
    plt.xlabel('Итерации');
    plt.ylabel('f1-score');
    plt.show();

    return None
```

### Полносвязная сеть - glove

```python
class GloveDataset(Dataset):
    def __init__(self, texts, labels, max_length=100, embedding_dim=100):
        super().__init__()
        self.texts = np.array(texts) if not isinstance(texts, np.ndarray) else texts
        self.labels = torch.LongTensor(labels) if not isinstance(labels, torch.Tensor) else labels
        self.max_length = max_length
        self.embedding_dim = embedding_dim
        
        print("Loading GloVe embeddings...")
        self.glove = load(f'glove-wiki-gigaword-{embedding_dim}')
        print("Embeddings loaded successfully!")
        
        self.pad_vector = torch.zeros(embedding_dim)
        
    def __len__(self):
        return len(self.texts)
    
    def __getitem__(self, idx):
        text = str(self.texts[idx])  
    
        words = text.split()[:self.max_length]
        
        vectors = []
        for word in words:
            try:
                vectors.append(torch.tensor(self.glove[word]))
            except KeyError:
                vectors.append(self.pad_vector)
        
        if len(vectors) < self.max_length:
            vectors += [self.pad_vector] * (self.max_length - len(vectors))
        
        vectors = vectors[:self.max_length]
        text_tensor = torch.stack(vectors)
        
        label = self.labels[idx]
        return text_tensor, label
```

```python
class AuthorClassifier(nn.Module):
    def __init__(self, max_length=100, embedding_dim=100, num_classes=3):
        super().__init__()
        self.net = nn.Sequential(
            nn.Flatten(),
            
            nn.Linear(max_length * embedding_dim, 2048),
            nn.BatchNorm1d(2048),
            nn.Dropout(0.6),
            nn.ELU(),
            
            nn.Linear(2048, 1024),
            nn.BatchNorm1d(1024),
            nn.Dropout(0.4),
            nn.ELU(),

            nn.Linear(in_features=1024, out_features=256),
            nn.BatchNorm1d(256),
            nn.Dropout(p=0.5),
            nn.ELU(),
            
            nn.Linear(256, num_classes)
        )


    def forward(self, x):
        if x.dim() == 2:
            x = x.unsqueeze(0)
        elif x.dim() == 3:
            pass
        else:
            raise ValueError(f"Unexpected input dimension: {x.dim()}")
            
        return self.net(x)
```

### Сверточная сеть - glove

```python
class GloVeDataset(Dataset):
    def __init__(self, X, y, max_len=200, embedding_name='glove-wiki-gigaword-100'):

        super().__init__()
        self.X = X
        self.y = y
        self.max_len = max_len
        
        print(f"Loading {embedding_name} embeddings...")
        self.embeddings = load(embedding_name)
        print(f"Embeddings loaded. Vector size: {self.embeddings.vector_size}")
        
        self.unk_vector = torch.zeros(self.embeddings.vector_size)
        
        self.y = torch.LongTensor(y.to_numpy())
        
    def __getitem__(self, idx):
        text = self.X.iloc[idx]
        tokens = text.split()[:self.max_len]
        
        word_vectors = []
        for token in tokens:
            try:
                word_vectors.append(torch.FloatTensor(self.embeddings[token]))
            except KeyError:
                word_vectors.append(self.unk_vector)
        
        if len(word_vectors) < self.max_len:
            padding = [torch.zeros(self.embeddings.vector_size)] * (self.max_len - len(word_vectors))
            word_vectors.extend(padding)
        
        return torch.stack(word_vectors), self.y[idx]
    
    def __len__(self):
        return len(self.X)

train_ds_glove = GloVeDataset(X_train, y_train, max_len=200)
valid_ds_glove = GloVeDataset(X_valid, y_valid, max_len=200)
test_ds_glove = GloVeDataset(X_test, y_test, max_len=200)

train_loader_glove = DataLoader(train_ds_glove, batch_size=16, shuffle=True)
valid_loader_glove = DataLoader(valid_ds_glove, batch_size=16, shuffle=True)
test_loader_glove =  DataLoader(test_ds_glove, batch_size=16, shuffle=True)
```

```python
class AuthorClassifier(nn.Module):
    def __init__(self, num_classes, embedding_dim=100, max_len=200, hidden_dim=128):  # Уменьшил hidden_dim
        super().__init__()
        
        self.conv_layers = nn.Sequential(
            nn.Conv1d(embedding_dim, 64, kernel_size=3, padding=1),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.Dropout(0.4),  
            nn.MaxPool1d(2),
            
            nn.Conv1d(64, 32, kernel_size=3, padding=1),
            nn.BatchNorm1d(32),
            nn.ReLU(),
            nn.Dropout(0.4),
            nn.AdaptiveAvgPool1d(1)
        )
        
        self.attention = nn.Sequential(
            nn.Linear(embedding_dim, 32),  
            nn.Tanh(),
            nn.Dropout(0.3),
            nn.Linear(32, 1, bias=False)
        )
        
        self.classifier = nn.Sequential(
            nn.Linear(32 + embedding_dim, hidden_dim),
            nn.BatchNorm1d(hidden_dim),
            nn.Dropout(0.6),  
            nn.ReLU(),
            nn.Linear(hidden_dim, num_classes)
        )
        
    def forward(self, x):
        batch_size = x.size(0)
        
        cnn_features = x.permute(0, 2, 1)
        cnn_features = self.conv_layers(cnn_features)
        cnn_features = cnn_features.view(batch_size, -1)
        
        attn_weights = F.softmax(self.attention(x), dim=1)
        attn_features = torch.sum(attn_weights * x, dim=1)
        
        combined = torch.cat([cnn_features, attn_features], dim=1)
        return self.classifier(combined)
```

```python
num_classes = y_train.nunique()
embedding_dim = train_ds_glove.embeddings.vector_size
max_len = train_ds_glove.max_len

model = AuthorClassifier(
    num_classes=num_classes,
    embedding_dim=embedding_dim,
    max_len=max_len,
    hidden_dim=128
)


device = 'cuda' if torch.cuda.is_available() else 'cpu'

optimizer = torch.optim.Adam(model.parameters(), lr=0.01, weight_decay=0.01)
loss_fn = torch.nn.CrossEntropyLoss()
metric_fn = torchmetrics.F1Score(task="multiclass", num_classes=num_classes).to(device)
epochs = 10000
scheduler = torch.optim.lr_scheduler.ExponentialLR(optimizer, gamma=0.95)
init = torch.nn.init.xavier_normal_
early_stopping = 10
grad_clip = None

history = train(
    model,
    train_loader_glove,
    valid_loader_glove,
    optimizer=optimizer,
    loss_fn=loss_fn,
    metric_fn=metric_fn,
    epochs=epochs,
    scheduler=scheduler,
    init=init,
    verbose=True,
    device=device,
    early_stopping=early_stopping,
    grad_clip=grad_clip
)
```

### LSTM - рекуррентные сети + tf-idf

```python
vectorizer = TfidfVectorizer(max_features=5000, ngram_range=(1, 2))

train_ds_tfidf = DatasetBooks(X_train, y_train, vectorizer, is_train=True, from_sparse_to_array=True)
valid_ds_tfidf = DatasetBooks(X_valid, y_valid, vectorizer, is_train=False, from_sparse_to_array=True)
test_ds_tfidf = DatasetBooks(X_test, y_test, vectorizer, is_train=False, from_sparse_to_array=True)

train_loader_tfidf = DataLoader(train_ds_tfidf, batch_size=16, shuffle=True)
valid_loader_tfidf = DataLoader(valid_ds_tfidf, batch_size=16, shuffle=True)
test_loader_tfidf = DataLoader(test_ds_tfidf, batch_size=16, shuffle=True)
```

```python
class AuthorLSTM(nn.Module):
    def __init__(self, input_size=5000, hidden_size=256, num_layers=1, num_classes=10, dropout=0.3):
        super().__init__()
        self.num_layers = num_layers
        self.hidden_size = hidden_size
        
        self.lstm = nn.LSTM(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=num_layers,
            batch_first=True,
            dropout=dropout if num_layers > 1 else 0
        )
        
        self.bn = nn.BatchNorm1d(hidden_size)
        
        self.dropout = nn.Dropout(dropout)
        
        self.fc = nn.Linear(hidden_size, num_classes)

 
    def forward(self, x):
        x = x.unsqueeze(1)  
        
        h0 = torch.zeros(self.num_layers, x.size(0), self.hidden_size).to(x.device)
        c0 = torch.zeros(self.num_layers, x.size(0), self.hidden_size).to(x.device)
        
        out, _ = self.lstm(x, (h0, c0))  
        
        out = out.squeeze(1)
        
        out = self.bn(out)
        out = self.dropout(out)
        
        out = self.fc(out)
        
        return out
```

```python
num_classes = y_train.nunique()

model = AuthorLSTM(
    input_size=5000, 
    hidden_size=64,
    num_layers=3,     
    num_classes=num_classes,
    dropout=0.3
)

optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
loss_fn = nn.CrossEntropyLoss()
metric_fn = torchmetrics.F1Score(task="multiclass", num_classes=num_classes).to(device)

history = train(
    model,
    train_loader_tfidf,
    valid_loader_tfidf,
    optimizer=optimizer,
    loss_fn=loss_fn,
    metric_fn=metric_fn,
    epochs=1000,
    device=device,
    early_stopping=50,
    grad_clip=3.0
)
```
